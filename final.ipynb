```python
import pandas as pd

# Load the dataset created by the analysis notebook before calculating KPIs.
main_df = pd.read_csv("olist_main_dataset.csv")
delivered = main_df[main_df["delivery_status"].isin(["Late", "On Time"])]
late_percentage = delivered["delivery_status"].eq("Late").mean() * 100
```

```python
print("""
OLIST E-COMMERCE DATA ANALYTICS PROJECT

Objective:
Analyze Olist e-commerce data to understand sales,
customers, products, payments, reviews, sellers,
and delivery performance.

Tools:
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Jupyter Notebook

Key Analysis:
1. Sales performance
2. Product category analysis
3. Customer analysis
4. Payment method analysis
5. Seller performance
6. Delivery performance
7. Customer reviews
8. Geographic analysis

Final KPIs:
""")

print("Total Sales Value:", round(main_df["total_cost"].sum(), 2))
print("Total Orders:", main_df["order_id"].nunique())
print("Total Customers:", main_df["customer_unique_id"].nunique())
print("Average Order Value:",
      round(main_df.groupby("order_id")["total_cost"].sum().mean(), 2))
print("Average Delivery Days:",
      round(main_df["delivery_days"].mean(), 2))
print("Average Review Score:",
      round(main_df["review_score"].mean(), 2))
print("Late Delivery %:",
      round(late_percentage, 2))
```
