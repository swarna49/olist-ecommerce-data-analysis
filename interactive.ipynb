```python
import pandas as pd

main_df = pd.read_csv("olist_main_dataset.csv")
```

```python
total_sales = main_df["total_cost"].sum()
total_orders = main_df["order_id"].nunique()
total_customers = main_df["customer_unique_id"].nunique()

print(total_sales)
print(total_orders)
print(total_customers)
```

```python
fig.add_trace(
    go.Indicator(
        mode="number",
        value=float(total_sales),
        title={"text": "Total Sales"},
    ),
    row=1,
    col=1,
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_orders,
        title={"text": "Total Orders"},
    ),
    row=1,
    col=2,
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_customers,
        title={"text": "Total Customers"},
    ),
    row=1,
    col=3,
)
```

```python
average_order_value = (
    main_df.groupby("order_id")["total_cost"]
    .sum()
    .mean()
)

average_delivery = main_df["delivery_days"].mean()

average_review = main_df["review_score"].mean()
```

```python
fig.add_trace(
    go.Indicator(
        mode="number",
        value=average_order_value,
        title={"text": "Average Order Value"}
    ),
    row=2,
    col=1
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=average_delivery,
        number={"suffix": " days"},
        title={"text": "Average Delivery"}
    ),
    row=2,
    col=2
)

fig.add_trace(
    go.Indicator(
        mode="number",
        value=average_review,
        number={"valueformat": ".2f"},
        title={"text": "Average Review Score"}
    ),
    row=2,
    col=3
)
```

```python
fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

dashboard = make_subplots(
    rows=3,
    cols=2,
    specs=[
        [{"type": "indicator"}, {"type": "indicator"}],
        [{"type": "xy"}, {"type": "xy"}],
        [{"type": "xy"}, {"type": "xy"}]
    ],
    subplot_titles=(
        "Total Sales",
        "Total Orders",
        "Monthly Sales",
        "Top Categories",
        "Payment Methods",
        "Delivery Status"
    )
)
```

```python
dashboard.add_trace(
    go.Indicator(
        mode="number",
        value=main_df["total_cost"].sum(),
        title={"text": "Total Sales"}
    ),
    row=1,
    col=1
)
```

```python
dashboard.add_trace(
    go.Indicator(
        mode="number",
        value=main_df["order_id"].nunique(),
        title={"text": "Total Orders"}
    ),
    row=1,
    col=2
)
```

```python
main_df["order_purchase_timestamp"] = pd.to_datetime(
    main_df["order_purchase_timestamp"]
)

monthly_revenue = (
    main_df
    .groupby(main_df["order_purchase_timestamp"].dt.to_period("M"))["total_cost"]
    .sum()
)

monthly_revenue.index = monthly_revenue.index.astype(str)
dashboard.add_trace(
    go.Scatter(
        x=monthly_revenue.index,
        y=monthly_revenue.values,
        mode="lines+markers",
        name="Sales",
    ),
    row=2,
    col=1,
)
dashboard.add_trace(
    go.Scatter(
        x=monthly_revenue.index,
        y=monthly_revenue.values,
        mode="lines+markers",
        name="Sales",
    ),
    row=2,
    col=1,
)
dashboard.show()
```

```python
category_revenue = (
    main_df.groupby("product_category_name_english")["total_cost"]
    .sum()
    .sort_values(ascending=False)
)

top_categories = category_revenue.head(10).sort_values()

dashboard.add_trace(
    go.Bar(
        x=top_categories.values,
        y=top_categories.index,
        orientation="h",
        name="Categories",
    ),
    row=2,
    col=2,
)

dashboard.add_trace(
    go.Bar(
        x=top_categories.values,
        y=top_categories.index,
        orientation="h",
        name="Categories"
    ),
    row=2,
    col=2
)
```

```python
payment_methods = (
    main_df["payment_count"]
    .dropna()
    .astype(int)
    .value_counts()
    .sort_index()
)

dashboard.add_trace(
    go.Bar(
        x=[f"{count} payment(s)" for count in payment_methods.index],
        y=payment_methods.values,
        name="Payment Count",
    ),
    row=3,
    col=1,
)
```

```python
delivery_counts = main_df["delivery_status"].value_counts()

dashboard.add_trace(
    go.Bar(
        x=delivery_counts.index,
        y=delivery_counts.values,
        name="Delivery"
    ),
    row=3,
    col=2
)
```

```python
dashboard.update_layout(
    title="Olist E-Commerce Analytics Dashboard",
    title_x=0.5,
    height=1000,
    showlegend=False
)

dashboard.show()
```
